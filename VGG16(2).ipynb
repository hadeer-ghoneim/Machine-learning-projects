```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import cv2
import tensorflow as tf
from tensorflow import keras
from tqdm import tqdm
```

```python
os.getcwd()
```

```python
train_path='dataset_animal/train'
val_path='dataset_animal/val'
testing_path='dataset_animal/testing'
```

```python
from keras.preprocessing.image import ImageDataGenerator

train_batch=ImageDataGenerator(preprocessing_function=tf.keras.applications.vgg16.preprocess_input).flow_from_directory(
directory=train_path,target_size=(224,224),batch_size=10)

val_batch=ImageDataGenerator(preprocessing_function=tf.keras.applications.vgg16.preprocess_input).flow_from_directory(
directory=val_path,target_size=(224,224),batch_size=10)

test_batch=ImageDataGenerator(preprocessing_function=tf.keras.applications.vgg16.preprocess_input).flow_from_directory(
directory=testing_path,target_size=(224,224),batch_size=10)
```

```python
vgg_16_model=tf.keras.applications.vgg16.VGG16()
```

```python
vgg_16_model.summary()
```

```python
model=keras.Sequential()
for layer in vgg_16_model.layers[:-1]:
    model.add(layer)
```

```python
for layer in model.layers:
    layer.trainable=False
```

```python
model.add(keras.layers.Dense(2,activation='Softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
results=model.fit(train_batch,validation_data=val_batch,steps_per_epoch=5,validation_steps=2,epochs=5,verbose=2)

# the steps_per_epoch  the number of batches you want to process per epoch.
# Validation Steps should reflect the number of batches you want to use for validation in each epoch.
```

```python
results.history['accuracy']
```

```python
#----convert probabilities to classes------
def transform_pred(pred):
  transformed_pred_list=[]
  for i in range(len(pred)):
    max_index=np.argmax(pred[i])
    transformed_pred_list.append(max_index)
  return transformed_pred_list
```
