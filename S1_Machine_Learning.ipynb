#Simple Linear Regression

```python
#1_import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#2_Data Frame
# Sample data: predicting house price based on size
data = {'size':[500, 750, 1000, 1250, 1500],
        'price': [150000, 200000, 250000, 300000, 350000] }
df = pd.DataFrame(data)
df
```

```python
#feature and target
# Define the feature (independent variable) and target (dependent variable)
x = df [['size']]
y = df ['price']
```

```python
#4_fit model
# Create and train the model
model = LinearRegression ()
model.fit(x,y)
```

```python
#5_predict
# Make predictions
predictions = model.predict(x)
predictions
```

```python
# 6_Calculate mean squared error and R-squared
from sklearn.metrics import mean_squared_error, r2_score
mse = mean_squared_error(y, predictions)
r2 = r2_score(y, predictions)
print(f'Mean Squared Error: {mse}')
print(f'R-squared: {r2}')
```

```python
print(f'Coefficients: {model.coef_}')
print(f'Intercept: {model.intercept_}')
```

```python
#7_plot
# Plot the results
plt.scatter(x ,y , color='blue' , label='actual data')
plt.plot(x, predictions, color='red', label='Linear Regression Line')
plt.xlabel('Size (sq ft)')
plt.ylabel('Price ($)')
plt.title('Simple Linear Regression: House Price Prediction')
plt.legend()
plt.show()
```

# Polynomial Regression

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
# Sample data: predicting disease progression based on time
data = {'Time': [1, 2, 3, 4, 5],
        'Progression': [1, 4, 9, 16, 25]}
df = pd.DataFrame(data)
```

```python
# Define the feature (independent variable) and target (dependent variable)
X = df[['Time']]
y = df['Progression']
```

```python
# Transform features to polynomial features (degree 2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
```

```python
# Create and train the model
model = LinearRegression()
model.fit(X_poly, y)
```

```python
# Make predictions
predictions = model.predict(X_poly)
```

```python
# Calculate mean squared error and R-squared
from sklearn.metrics import mean_squared_error, r2_score
mse = mean_squared_error(y, predictions)
r2 = r2_score(y, predictions)
print(f'Mean Squared Error: {mse}')
print(f'R-squared: {r2}')
```

```python
print(f'Coefficients: {model.coef_}')
print(f'Intercept: {model.intercept_}')
```

```python
# Plot the results
plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, predictions, color='red', label='Polynomial Regression Line')
plt.xlabel('Time')
plt.ylabel('Progression')
plt.title('Polynomial Regression: Disease Progression Prediction')
plt.legend()
plt.show()
```
