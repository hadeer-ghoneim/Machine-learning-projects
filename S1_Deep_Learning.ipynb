# **Mathematical Representation of a Perceptron**



1.   **Perceptron forward pass:** Compute the weighted sum of the inputs.
2.   **Apply activation function:** Use the sigmoid function to control the output.
3.   **Compute cost function:** Calculate the error between predicted and actual values.
4.   **Weight update:** Adjust the weights based on the error.



Import NumPy

```python
import numpy as np
```

Sigmoid activation function

```python
# Sigmoid activation function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

Derivative of the sigmoid function for backpropagation

```python
# Derivative of the sigmoid function for backpropagation
def sigmoid_derivative(z):
    return sigmoid(z) * (1 - sigmoid(z))
```

Perceptron forward pass: calculate weighted sum

```python
# Perceptron forward pass: calculate weighted sum
def perceptron(inputs, weights, bias):
    # Weighted sum (dot product of inputs and weights + bias)
    z = np.dot(inputs, weights) + bias
    return z
```

Apply activation function (sigmoid in this case)

```python
# Apply activation function (sigmoid in this case)
def activate(z):
    return sigmoid(z)
```

Cost function: Mean Squared Error (MSE)

```python
# Cost function: Mean Squared Error (MSE)
def compute_cost(y_pred, y_actual):
    return np.mean((y_pred - y_actual) ** 2)
```

Update weights and bias

```python
# Update weights and bias
def update_weights(inputs, weights, bias, y_pred, y_actual, learning_rate=0.01):
    # Calculate the error
    error = y_pred - y_actual
    # Calculate the gradient for weights (using the derivative of the cost function)
    weights_gradient = np.dot(inputs.T, error * sigmoid_derivative(y_pred))
    # Calculate the gradient for bias
    bias_gradient = np.sum(error * sigmoid_derivative(y_pred))

    # Update weights and bias
    weights -= learning_rate * weights_gradient
    bias -= learning_rate * bias_gradient

    return weights, bias
```

Example usage

```python
# Example usage
inputs = np.array([[0.5, 0.3], [0.2, 0.1], [0.6, 0.4]])  # 3 samples, 2 features
weights = np.random.rand(2)  # Random initial weights (2 features)
bias = np.random.rand(1)  # Random initial bias

y_actual = np.array([1, 0, 1])  # Example target values
```

Forward pass

```python
# Forward pass
z = perceptron(inputs, weights, bias)
y_pred = activate(z)
```

Compute cost

```python
# Compute cost
cost = compute_cost(y_pred, y_actual)
print(f"Cost: {cost}")
```

Update weights and bias

```python
# Update weights and bias
weights, bias = update_weights(inputs, weights, bias, y_pred, y_actual)
print(f"Updated Weights: {weights}")
print(f"Updated Bias: {bias}")
```
