```python
tform = transform.SimilarityTransform(rotation = np.deg2rad(30))
tform
```

```python
im_ = transform.warp(image = im,
                     inverse_map = tform.inverse)
imshow(im_)
plt.show()
```

```python
theta = np.deg2rad(30)
center = np.array(im.shape) // 2

A = np.array([[1, 0, -center[1]],
              [0, 1, -center[0]],
              [0, 0, 1]])

B = np.array([[np.cos(theta), -np.sin(theta), 0],
              [np.sin(theta), np.cos(theta), 0],
              [0, 0, 1]])

C = np.array([[1, 0, center[1]],
              [0, 1, center[0]],
              [0, 0, 1]])

tform = transform.SimilarityTransform(matrix = np.dot(np.dot(C, B), A))
im_ = transform.warp(image = im,
                     inverse_map = tform.inverse)
imshow(im_)
plt.show()
print(tform)
```

```python
shift_center_tform = transform.SimilarityTransform(translation = (-center[1], -center[0]))
rotate_tform = transform.SimilarityTransform(rotation = np.deg2rad(30))
inv_shift_center_tform = shift_center_tform.inverse
tform = shift_center_tform + rotate_tform + inv_shift_center_tform
tform
```

```python
s = 8
```

```python
w = np.ones(shape = [s, s])
w = w / w.sum()
w
```

```python
im_ = np.zeros(shape = np.array(im.shape) // s, dtype = 'uint8')
im_.shape
```

```python
for i in range(im_.shape[0]):
    for j in range(im_.shape[1]):
        i_start, j_start = s * i, s * j
        i_end, j_end = i_start + s, j_start + s
        r = im[i_start : i_end, j_start : j_end]
        im_[i, j] = np.uint8((r * w).sum())
        
imshow(im_)
plt.show()
```
